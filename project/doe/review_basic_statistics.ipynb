# Population vs Sample

Compare population parameters with sample statistics.

````python
import os
os.environ.setdefault("MPLCONFIGDIR", "/tmp/matplotlib")

import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Markdown, display

np.random.seed(7)

population = np.random.normal(loc=17.0, scale=0.35, size=10000)
sample = np.random.choice(population, size=10, replace=False)
````

````python
comparison = [
    ("Population mean", population.mean()),
    ("Sample mean", sample.mean()),
    ("Population standard deviation", population.std(ddof=0)),
    ("Sample standard deviation", sample.std(ddof=1)),
]

table = "| Statistic | Value |\n|---|---:|\n"
for name, value in comparison:
    table += f"| {name} | {value:.6f} |\n"

display(Markdown(table))
````

````python
fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(population, bins=25, edgecolor="black", alpha=0.75)
ax.axvline(population.mean(), color="black", linestyle="--", label="Population mean")
ax.axvline(sample.mean(), color="red", linestyle="--", label="Sample mean")
ax.set_title("Population Distribution")
ax.set_xlabel("Value")
ax.set_ylabel("Frequency")
ax.legend()
plt.show()
````

````python
fig, ax = plt.subplots(figsize=(7, 2))
ax.scatter(sample, np.zeros_like(sample), s=70)
ax.axvline(sample.mean(), color="red", linestyle="--", label="Sample mean")
ax.set_yticks([])
ax.set_title("Sample Dot Plot")
ax.set_xlabel("Value")
ax.legend()
plt.show()
````

````python
fig, ax = plt.subplots(figsize=(4, 4))
ax.boxplot(sample)
ax.set_title("Sample Box Plot")
ax.set_ylabel("Value")
plt.show()
````

## Core Code to Remember

```python
# Create population and sample
population = np.random.normal(loc=17.0, scale=0.35, size=10000)
sample = np.random.choice(population, size=10, replace=False)

# Sample statistics: use ddof=1
sample.mean()
sample.var(ddof=1)
sample.std(ddof=1)

# Population parameters: use ddof=0
population.mean()
population.var(ddof=0)
population.std(ddof=0)

# Histogram
ax.hist(population, bins=25, edgecolor="black")

# Dot plot
ax.scatter(sample, np.zeros_like(sample))

# Box plot
ax.boxplot(sample)

# Mean reference line
ax.axvline(sample.mean(), color="red", linestyle="--")
```
